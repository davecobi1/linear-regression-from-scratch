```python
x = np.array(df['Hours_Studied'])
y = np.array(df['Exam_Score'])
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state = 42, test_size = 0.2)
```

### Building the model

```python
def model(w, b, x):
    return w*x + b
```

```python
# Initializing some parameters
w = 0
b = 0
alpha = 0.001
iterations = 100000
m = x_train.shape[0]
iters, costs = [], []
```

```python
for iteration in range(iterations):
    total_squared_error = 0
    b_error_term = 0
    w_error_term = 0
    for row in range(m):
        y_pred = model(w,b, x_train[row])
        total_squared_error += (y_pred - y_train[row])**2
        w_error_term += (y_pred - y_train[row])*x_train[row]
        b_error_term += (y_pred - y_train[row])
    w = w - alpha*(1/m)*w_error_term
    b = b - alpha*(1/m)*b_error_term
    cost = 0.5*(1/m)*total_squared_error

    if (iteration % 100 == 0):
        print(f"Iteration:  {iteration}")
        print(f"Cost:  {cost}")
        iters.append(iteration)
        costs.append(cost)
        
```

```python
plt.title("A plot showing the fall in the cost with each iteration")
plt.plot(iters , costs)
plt.show()
```

```python
print(f"Model Parameters:")
print(f"Weight: {w}")
print(f"Bias: {b}")
```

```python
y_predicted = model(w,b, x_test)
# print(type(x_test))
# print(y_predicted[0:10], y_test[0:10])
r2 = r2_score(y_test, y_predicted)
mae = mean_absolute_error(y_test, y_predicted)
mse = mean_squared_error(y_test, y_predicted)

print(f"Mean Squared Error: {mse}")
print(f"Mean Absolute Error: {mae}")
print(f"R Squared: {r2}")
```
